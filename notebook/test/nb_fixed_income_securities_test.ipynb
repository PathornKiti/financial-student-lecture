## Question 1 — Lottery payments

A lottery advertises a prize of **$10 million**, but it is actually paid as **20 annual payments of $500,000**, with the **first payment immediately**.

We are asked for the **present value** at **10% annual compounding**, reported in **$millions**.

Because the first payment is immediate, this is an **annuity due**.

### Formula

The present value is

$$
PV = 500{,}000 \sum_{i=0}^{19} \frac{1}{(1.10)^i}
$$

Since the first payment is at time 0, the sum starts at $i=0$.

You can also write the annuity-due formula as

$$
PV = 500{,}000 \cdot \frac{1-(1.10)^{-20}}{0.10} \cdot 1.10
$$


```python
payment = 500_000
r = 0.10
n = 20

pv_lottery = payment * sum(1 / ((1 + r) ** i) for i in range(n))
pv_lottery_millions = pv_lottery / 1_000_000

pd.DataFrame(
    {
        "quantity": ["PV in dollars", "PV in millions"],
        "value": [pv_lottery, pv_lottery_millions],
    }
)
```

### Answer

$$
PV = 4.68246 \text{ million dollars}
$$

Rounded to two decimal places:

$$
\boxed{4.68}
$$

So the correct answer is **4.68 million**, not 5.15.


## Question 2 — Sunk costs

A couple has already committed to a 6-month apartment lease at **$1,000 per month** and has paid a **$1,000 refundable deposit**.  
The next day they find another apartment for **$900 per month**, also with a **$900 refundable deposit**.

They will stay only 6 months. The interest rate is **12% per month compounded monthly**.

The key lesson is that this is a **sunk cost** problem. The original deposit decision has already been made, so the correct comparison is based on the cash flows going forward.

Using the setup shown in your test explanation, compare the two NPVs.

### Stay in current apartment

$$
C_1 = -\sum_{i=0}^{5} \frac{1000}{1.12^i} + \frac{1000}{1.12^6}
$$

### Switch to the new apartment

$$
C_2 = -900 - \sum_{i=0}^{5} \frac{900}{1.12^i} + \frac{900}{1.12^6}
$$

Choose the alternative with the **higher** NPV (that is, the less negative one).


```python
monthly_r = 0.12

C1 = -sum(1000 / ((1 + monthly_r) ** i) for i in range(6)) + 1000 / ((1 + monthly_r) ** 6)
C2 = -900 - sum(900 / ((1 + monthly_r) ** i) for i in range(6)) + 900 / ((1 + monthly_r) ** 6)

pd.DataFrame(
    {
        "alternative": ["Stay", "Switch"],
        "NPV": [C1, C2],
    }
)
```

### Answer

From the calculation,

$$
C_1 \approx -4098
$$

and

$$
C_2 \approx -4588
$$

Since staying has the **higher NPV**, the couple should

$$
\boxed{\text{Stay}}
$$


## Question 3 — Relation between spot and discount rates

Given annual-compounded spot rates:

- $s_1 = 6.3\%$
- $s_2 = 6.9\%$

find the discount factor $d(0,2)$.

### Formula

With annual compounding,

$$
d(0,t) = \frac{1}{(1+s_t)^t}
$$

So here,

$$
d(0,2) = \frac{1}{(1+0.069)^2}
$$


```python
s2 = 0.069
d02 = 1 / ((1 + s2) ** 2)
d02
```

### Answer

$$
d(0,2) = 0.8751
$$

Rounded to three decimal places:

$$
\boxed{0.875}
$$


## Question 4 — Relation between spot and forward rates

Given:

- $s_1 = 6.3\%$
- $s_2 = 6.9\%$

find the **1-year forward rate starting in 1 year**, denoted $f_{1,2}$, assuming annual compounding.

### Formula

The no-arbitrage relation is

$$
(1+s_2)^2 = (1+s_1)(1+f_{1,2})
$$

So,

$$
f_{1,2} = \frac{(1+s_2)^2}{1+s_1} - 1
$$


```python
s1 = 0.063
s2 = 0.069

f12 = ((1 + s2) ** 2) / (1 + s1) - 1
f12
```

### Answer

$$
f_{1,2} = 0.075 = 7.5\%
$$

Rounded to one decimal place as a percentage:

$$
\boxed{7.5\%}
$$


## Question 5 — Forward contract on a stock

The current stock price is

$$
S_0 = 400
$$

The stock pays **no dividend**. The interest rate is **8% per year compounded quarterly**.  
Delivery is in **9 months**, which is **3 quarters**.

### Formula

With discrete compounding, the forward price is

$$
F_0 = S_0(1 + r/m)^{mT}
$$

Here:

- $r = 0.08$
- $m = 4$
- $T = 0.75$
- so $mT = 3$

Thus,

$$
F_0 = 400(1+0.08/4)^3
$$


```python
S0 = 400
r = 0.08
m = 4
T = 9/12

forward_price = S0 * (1 + r / m) ** (m * T)
forward_price
```

### Answer

$$
F_0 = 424.4832
$$

Rounded to two decimal places:

$$
\boxed{424.48}
$$


## Question 6 — Bounds using different lending and borrowing rates

Suppose:

- borrowing rate: $r_B = 10\%$
- lending rate: $r_L = 8\%$

We want the difference between the **upper bound** and **lower bound** on the value of a **perpetuity** that pays

$$
A = 10{,}000
$$

per year.

### Bounds

For a perpetuity:

- Upper bound uses the **lending rate**
- Lower bound uses the **borrowing rate**

So:

$$
\text{Upper bound} = \frac{A}{r_L} = \frac{10{,}000}{0.08} = 125{,}000
$$

$$
\text{Lower bound} = \frac{A}{r_B} = \frac{10{,}000}{0.10} = 100{,}000
$$

Difference:

$$
125{,}000 - 100{,}000 = 25{,}000
$$


```python
A = 10_000
rL = 0.08
rB = 0.10

upper_bound = A / rL
lower_bound = A / rB
difference = upper_bound - lower_bound

pd.DataFrame(
    {
        "quantity": ["upper_bound", "lower_bound", "difference"],
        "value": [upper_bound, lower_bound, difference],
    }
)
```

### Answer

$$
\boxed{25{,}000}
$$


## Question 7 — Value of a forward contract at an intermediate time

We hold a forward contract on a stock:

- entered **6 months ago**
- original maturity was **1 year**
- so there are **6 months remaining**
- stock price 6 months ago: $S_0 = 100$
- current stock price: $S_t = 125$
- interest rate: $r = 10\%$ compounded semi-annually

We want the **current value** of the forward contract.

### Step 1: Original delivery price

At inception, the forward price was

$$
F_0 = S_0(1+r/2)^2 = 100(1.05)^2 = 110.25
$$

### Step 2: Current forward price for the remaining 6 months

With 6 months left,

$$
F_t = S_t(1+r/2) = 125(1.05) = 131.25
$$

### Step 3: Current value of the long forward

The value is the present value of the difference between the current forward price and the original delivery price:

$$
f_t = (F_t - F_0) \cdot d(t,T)
$$

Since there is 6 months left and semiannual compounding,

$$
d(t,T) = \frac{1}{1+r/2} = \frac{1}{1.05}
$$

Therefore,

$$
f_t = \frac{131.25 - 110.25}{1.05} = 20
$$


```python
S_init = 100
S_t = 125
r = 0.10

F0_original = S_init * (1 + r / 2) ** 2
F_t_current = S_t * (1 + r / 2)
f_t = (F_t_current - F0_original) / (1 + r / 2)

pd.DataFrame(
    {
        "quantity": ["original_forward_price", "current_forward_price", "forward_value_now"],
        "value": [F0_original, F_t_current, f_t],
    }
)
```

### Answer

$$
\boxed{20}
$$
